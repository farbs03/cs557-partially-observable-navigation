```python
import gymnasium as gym
import popgym
from popgym.envs.labyrinth_escape import LabyrinthEscapeEasy
from sb3_contrib import RecurrentPPO
from stable_baselines3.common.env_util import make_vec_env
from gymnasium.wrappers import FlattenObservation
```

```python
def make_env():
    env = LabyrinthEscapeEasy()
    env = FlattenObservation(env) # Flatten observation so MLP feature extractor can take it
    return env
```

```python
# Vectorize environment for stability and faster data collection
vec_env = make_vec_env(make_env, n_envs=4)
```

```python
print(vec_env.observation_space)
print(vec_env.action_space)
```

```python
policy_type = "MlpLstmPolicy"
```

```python
model = RecurrentPPO(
    policy=policy_type,
    env=vec_env,
    verbose=1,
    n_steps=256,
    policy_kwargs=dict(
        lstm_hidden_size=128,
        n_lstm_layers=1,    
    ),
    tensorboard_log="./tensorboard_logs"
)
```

```python
model.learn(total_timesteps=200000)
model.save("./models/recurrent_ppo_labyrinth_easy_mlplstm")
vec_env.close()
```
